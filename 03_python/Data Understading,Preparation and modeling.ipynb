# Data Dleaning

```python
df = csb.copy()
```

```python
# Clean and normalize column names
df.columns = df.columns.str.lower().str.replace(' ','_')
df = df.rename(columns = {'purchase_amount_(usd)' : 'purchase_amount'})
df.columns
```

```python
df = df.astype({'customer_id':'int16','age':'int8','gender':'category','category':'category',
    'size':'category','season':'category','subscription_status':'category','shipping_type':'category',
         'discount_applied':'category','payment_method':'category','frequency_of_purchases':'category'})
```

```python
df.info()
```

**merory usage decrease from 548.6+ KB to 260.9+ KB**

# Feature Engineering

```python
def grouping_age(age):
    """Categorize age values into human-readable segments."""
    if 18 <= age <= 25:
        return 'Young_Adult'
    elif 26 <= age <= 35:
        return 'Adult'
    elif 36 <= age <= 45:
        return 'Senior_Adult'
    elif 46 <= age <= 60:
        return 'Middle_Aged'
    elif age > 60:
        return 'Senior'
    else:
        return 'Teenager'

df['age_group'] = df['age'].apply(grouping_age)
df[['age','age_group']]
```

```python
df['frequency_of_purchases'].value_counts()
```

```python
frequency_mapping = {
            'Fortnightly': 14,
            'Weekly': 7,
            'Monthly': 30,
            'Quarterly': 90,
            'Bi-Weekly': 14,
            'Annually': 365,
            'Every 3 Months': 90
        }
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases','purchase_frequency_days']]
```

```python
(df['promo_code_used'] == df['discount_applied']).sum()
```

```python
# Drop 'promo_code_used' column because it's value is similar with 'discount_applied', Thus it's create redundency
df.drop('promo_code_used',axis=1,inplace=True)
```
