```python
df = csb.copy()
```

```python
# ---------------------------------------
# Convert categorical columns
# ---------------------------------------

category_columns = [
    "Gender",
    "Item Purchased",
    "Category",
    "Location",
    "Size",
    "Color",
    "Season",
    "Subscription Status",
    "Shipping Type",
    "Discount Applied",
    "Promo Code Used",
    "Payment Method",
    "Frequency of Purchases"
]

for col in category_columns:
    df[col] = df[col].astype("category")


# ---------------------------------------
# Optimize numeric columns
# ---------------------------------------

df["Customer ID"] = df["Customer ID"].astype("int32")

df["Age"] = df["Age"].astype("int16")

df["Purchase Amount (USD)"] = df["Purchase Amount (USD)"].astype("int32")

df["Previous Purchases"] = df["Previous Purchases"].astype("int32")
```

```python
# Clean and normalize column names
df.columns = df.columns.str.lower().str.replace(' ','_')
df = df.rename(columns = {'purchase_amount_(usd)' : 'purchase_amount'})
df.columns
```

```python
df.info()
```

**merory usage decrease from 548.6+ KB to 260.9+ KB**

```python
def grouping_age(age):
    """Categorize age values into human-readable segments."""
    if 18 <= age <= 25:
        return 'Young_Adult'
    elif 26 <= age <= 35:
        return 'Adult'
    elif 36 <= age <= 45:
        return 'Senior_Adult'
    elif 46 <= age <= 60:
        return 'Middle_Aged'
    elif age > 60:
        return 'Senior'
    else:
        return 'Teenager'

df['age_group'] = df['age'].apply(grouping_age)
df[['age','age_group']]
```

```python
df['frequency_of_purchases'].value_counts()
```

```python
frequency_mapping = {
            'Fortnightly': 14,
            'Weekly': 7,
            'Monthly': 30,
            'Quarterly': 90,
            'Bi-Weekly': 14,
            'Annually': 365,
            'Every 3 Months': 90
        }
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases','purchase_frequency_days']]
```
