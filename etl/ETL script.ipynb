```python
import pandas as pd
from sqlalchemy import create_engine, text


engine = create_engine("postgresql+psycopg2://postgres:721401@localhost:5432/projects")

schema_name = "customers_shopping_behavior"

with engine.begin() as connection:
    connection.execute(
        text(f"CREATE SCHEMA IF NOT EXISTS {schema_name}")
    )

df = pd.read_csv("dataset/updated_customer_shopping_behaviour_data.csv")

df.to_sql(
    name= "shopping_behavior",
    con=engine,
    schema=schema_name,
    if_exists="replace",
    index=False,
    method="multi"
 )

print(f"Uploaded successfully: {schema_name}.{table_name}\n")

print("ETL completed successfully!")
```
